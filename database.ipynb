```python
# Install the required libraries
!pip install google-api-python-client

from googleapiclient.discovery import build
import json
import time

# Replace with your actual API key
API_KEY = "AIzaSyBoXHNX8LL63roA6bU-cTfr3OO-zOQZLS4"  # 🔹 Replace with your valid API Key

# Initialize YouTube API client
youtube = build("youtube", "v3", developerKey=API_KEY)

# Search Query: "Top Marketing Strategies in Tech Industry"
query = "Top Marketing Strategies in Tech Industry"

# Function to get video details
def get_video_details(video_ids):
    """
    Fetch video details including title, description, channel, upload date, views, likes, and comments.
    """
    video_details = []
    for i in range(0, len(video_ids), 50):  # API allows max 50 video IDs per request
        video_id_batch = video_ids[i:i+50]
        response = youtube.videos().list(
            id=",".join(video_id_batch),
            part="snippet,statistics"
        ).execute()

        for item in response.get("items", []):
            stats = item.get("statistics", {})
            snippet = item.get("snippet", {})

            video_data = {
                "video_id": item["id"],
                "url": f"https://www.youtube.com/watch?v={item['id']}",
                "title": snippet.get("title", "N/A"),
                "description": snippet.get("description", "N/A"),
                "channel_name": snippet.get("channelTitle", "N/A"),
                "upload_date": snippet.get("publishedAt", "N/A"),
                "views": int(stats.get("viewCount", 0)),
                "likes": int(stats.get("likeCount", 0)),
                "comments": int(stats.get("commentCount", 0))
            }
            video_details.append(video_data)

    return video_details

# Collect video IDs
videos = []
nextPageToken = None

while True:
    # Search for videos
    search_response = youtube.search().list(
        q=query,
        part="id,snippet",
        maxResults=50,  # Max per API call
        type="video",
        order="viewCount",  # Sort by most viewed videos
        pageToken=nextPageToken
    ).execute()

    # Extract video IDs
    video_ids = [item["id"]["videoId"] for item in search_response.get("items", []) if "videoId" in item["id"]]

    if not video_ids:
        break  # No more videos

    # Fetch video details
    videos.extend(get_video_details(video_ids))

    # Get the next page token (for pagination)
    nextPageToken = search_response.get("nextPageToken")
    if not nextPageToken:
        break  # No more pages left

    time.sleep(1)  # Avoid hitting rate limits

# Save data to JSON file
with open("marketing_videos_metadata.json", "w") as f:
    json.dump(videos, f, indent=2)

print(f"✅ Collected metadata for {len(videos)} videos and saved to youtube_marketing_videos_metadata.json")
```

```python
import yt_dlp
import os

AUDIO_DIR = "audio_files"
os.makedirs(AUDIO_DIR, exist_ok=True)  # Create directory if not exists

def download_audio(video_url, output_path):
    ydl_opts = {
        "format": "bestaudio/best",
        "postprocessors": [{
            "key": "FFmpegExtractAudio",
            "preferredcodec": "mp3",
            "preferredquality": "192",
        }],
        "outtmpl": output_path,  # Save audio with video ID
    }

    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            ydl.download([video_url])
        print(f"✅ Audio downloaded: {output_path}")
    except Exception as e:
        print(f"❌ Error downloading {video_url}: {e}")

# Download all videos
for video in videos:
    video_url = video["url"]
    audio_file = os.path.join(AUDIO_DIR, f"{video['video_id']}.mp3")
    download_audio(video_url, audio_file)
```

```python
import os
import json
import whisper

# ✅ Load Whisper Model
try:
    model = whisper.load_model("large")
    print("✅ Whisper model loaded successfully!")
except Exception as e:
    print(f"❌ Error loading model: {e}")

# ✅ Define Paths in Google Colab
AUDIO_DIR = "/content/audio_files"
TRANSCRIPT_DIR = "/content/transcripts"
OUTPUT_PATH = "/content/marketing_videos_with_transcripts.json"

# ✅ Ensure transcript directory exists
os.makedirs(TRANSCRIPT_DIR, exist_ok=True)

def transcribe_audio(audio_path):
    """
    ✅ Converts audio to text using Whisper API.
    """
    try:
        # Load and transcribe audio using Whisper
        print(f"🔹 Transcribing {audio_path} using Whisper...")
        result = model.transcribe(audio_path)
        transcript = result["text"]

        # ✅ Save transcript as a text file
        transcript_file = os.path.join(TRANSCRIPT_DIR, f"{os.path.basename(audio_path)}.txt")
        with open(transcript_file, "w") as f:
            f.write(transcript)

        print(f"✅ Transcription saved: {transcript_file}")
        return transcript

    except Exception as e:
        print(f"❌ Error transcribing {audio_path}: {e}")
        return ""

# ✅ Load video metadata (if needed)
videos = []

# ✅ Transcribe all audio files and update dataset
for audio_file in sorted(os.listdir(AUDIO_DIR)):  # Sort for consistency
    if audio_file.endswith(".mp3"):
        audio_path = os.path.join(AUDIO_DIR, audio_file)
        transcript = transcribe_audio(audio_path)

        # Extract video_id from filename (assuming it's saved as "video_id.mp3")
        video_id = os.path.splitext(audio_file)[0]

        # Append data to list
        videos.append({
            "video_id": video_id,
            "transcript": transcript
        })

# ✅ Save updated metadata with transcripts
with open(OUTPUT_PATH, "w") as f:
    json.dump(videos, f, indent=2)

print(f"✅ All audio files transcribed and saved successfully at {OUTPUT_PATH}")
```
